### I.II Explanation of Models

#### 1. Random Forest (Primary Model):

**Overview:**  
Random Forest is a machine learning algorithm that works by building a large number of decision trees during training and then combining their predictions to make the final output. Each tree is trained on a slightly different sample of the data, which is known as bootstrapping, and only a subset of features is used when splitting nodes. This randomness adds variety among the trees and helps avoid the problem of all trees making the same mistake.

**Why It’s Used:**  
I chose to use Random Forest in this project mainly because of its ability to handle many features at once without easily overfitting. Since stock market data can be noisy and sometimes unpredictable, Random Forest’s strategy of averaging many trees helps smooth out random patterns that might lead a single decision tree to memorize the training data. This way, the model becomes more general and better at making accurate predictions on new, unseen data. Also, it doesn’t require heavy parameter tuning to start working well, which is a plus for someone like me still learning.

**Application in This Project:**  
In this project, the Random Forest model is trained using various engineered features from historical Tesla stock data. These features include daily prices, volume, and technical indicators like rolling averages and momentum scores. The goal is to predict whether the price will go up or down the next day. What makes Random Forest great here is that it gives not only a prediction but also a probability, so I can get a sense of how confident the model is in its choice. That extra confidence score is useful when deciding whether to take action or not. The model’s ensemble structure also brings consistency, which is really important in such a volatile environment as the stock market.

**Advantages:**

#### 4. XGBoost:

**Overview:**  
XGBoost stands for eXtreme Gradient Boosting.
```python
# Feature engineering
def add_technical_indicators(df):
    """
    Adds technical indicators to the DataFrame.

    Args:
        df (pd.DataFrame): Stock data with a 'Close' column.

    Returns:
        pd.DataFrame: DataFrame with additional technical indicator columns.
    """

    horizons = [2,5,60,260,780] #keeping in mind the 5 working days
    global predictors
    predictors = []

    df["The_day_after"] = df["Close"].shift(-1)
    df["Did_it_increase (Target)"] = (df["The_day_after"]>df["Close"]).astype(int)

    for horizon in horizons:
        rolling_averages = df.rolling(horizon).mean()

        ratio_column = f"Close_Ratio_{horizon}"
        df[ratio_column] = df["Close"] / rolling_averages["Close"]

        trend_column = f"Trend_{horizon}"
        df[trend_column] = df.shift(1).rolling(horizon).sum()["Did_it_increase (Target)"]

        predictors+= [ratio_column, trend_column]



    df["LRDA"] = np.log(1 + df['Close'].pct_change()).shift(-1) #log_returns_of_the_day_after

    # Rolling averages
    df["SMA_20"] = df["Close"].rolling(20).mean()
    df["SMA_50"] = df["Close"].rolling(50).mean()

    # RSI Relative Strength
    delta = df["Close"].diff(1)
    gain = (delta.where(delta > 0, 0)).rolling(14).mean()
    loss = (-delta.where(delta < 0, 0)).rolling(14).mean()
    rs = gain / loss
    df["RSI"] = 100 - (100 / (1 + rs))

    # Bollinger Bands
    df["BB_Upper"] = df["SMA_20"] + (df["Close"].rolling(20).std() * 2)
    df["BB_Lower"] = df["SMA_20"] - (df["Close"].rolling(20).std() * 2)

    # Volatility (Rolling Std Dev)
    df["Volatility"] = df["Close"].rolling(20).std()

    # Lag features
    df["Lag_1"] = df["Close"].shift(1)
    df["Lag_2"] = df["Close"].shift(2)
    predictors += ["SMA_20", "SMA_50", "RSI", "BB_Upper", "BB_Lower", "Volatility", "Lag_1", "Lag_2","Stock Splits"]

    return df
```

```python
# Create a Random Forest Classifier with specified hyperparameters
model = RandomForestClassifier(n_estimators=1500, min_samples_split=50, random_state=1)
```

### Explanation of Parameters

The parameters selected for the model help shape its behavior and performance in important ways.

**`n_estimators=1500`**  
This defines the number of decision trees that the Random Forest will build. Having a high number of trees generally improves accuracy because it reduces variance and helps the model generalize better to unseen data. However, going too high might slow down training and increase computational costs without giving a major boost in performance.

**`min_samples_split=50`**  
This sets the minimum number of samples required to split a node in any of the trees. A higher value like 50 means the tree won’t split unless there are at least 50 samples in a node, which helps prevent the model from becoming too complex or overfitting the training data. It acts as a form of regularization and encourages broader, more general splits.

**`random_state=1`**  
This ensures that the model produces the same results every time it’s run by setting a fixed seed for the random number generator. Without it, the tree-building process would vary each time, making it harder to reproduce results.

### Train-Test Split and Prediction Process

The dataset was divided into two parts: a **training set** and a **test set**. Specifically, we used `train = df.iloc[:-100]` and `test = df.iloc[-100:]`, which means the model is trained on all the data except the most recent 100 rows. The last 100 entries are kept aside for testing. This separation is important because it avoids any kind of data leakage — the model never gets to “see” the future data it's later tested on, which keeps the evaluation fair and realistic.

The actual training of the model is triggered with the line `model.fit(train[predictors], train["Did_it_increase (Target)"])`. This fits the model to the training data using the selected predictor columns (features), and the target column (which tells whether the stock went up or not the next day).

We then use a function called `predict()` to handle the process of making predictions. This function trains the model on the training data, then uses `predict_proba()` to return probabilities for each class on the test data — specifically, the second column of the result `[ :, 1 ]` gives the probability that the stock will increase. If that probability is 60% or higher, it is converted into a prediction of 1 (meaning: yes, it will go up). Otherwise, it's labeled as 0 (meaning: no increase).

Finally, the predicted values and the actual values are combined into one DataFrame so that we can compare them directly and evaluate how well the model performed.


```python
def predict(train, test, predictors, model):
    """
    Makes predictions using the trained model.

    Args:
        train (pd.DataFrame): Training data.
        test (pd.DataFrame): Test data.
        predictors (list): List of feature column names.
        model (sklearn.base.BaseEstimator): Trained model.

    Returns:
        pd.DataFrame: DataFrame with actual and predicted values.
    """
    model.fit(train[predictors], train["Did_it_increase (Target)"])
    preds = model.predict_proba(test[predictors])[:,1] #predict_proba predicts class probabilities
    #notice we used the test and not train
    #second coloumn is the probability of the stock going up
    preds[preds >=.6] = 1 #again probability
    preds[preds <.6] = 0 #anything below 60% certainity becomes a 0

    preds = pd.Series(preds, index=test.index, name="Predictions")
    combined = pd.concat([test["Did_it_increase (Target)"], preds], axis=1)
    return combined
```

```python
def backtest(data, model, predictors, start=500, step=100): #make sure the selected start and step is relevant to our dataset otherwise there will be nothing concatenate
    all_predictions = []
    """
    Backtests the model over time.

    Args:
        data (pd.DataFrame): Full dataset.
        model (sklearn.base.BaseEstimator): Model to test.
        predictors (list): List of feature column names.
        start (int): Starting index for backtesting.
        step (int): Step size for backtesting.

    Returns:
        pd.DataFrame: DataFrame of all predictions during backtesting.
    """


    for i in range(start, data.shape[0], step):
        train = data.iloc[0:i].copy()
        test = data.iloc[i:(i+step)].copy()
        predictions = predict(train, test, predictors, model)
        all_predictions.append(predictions)

    return pd.concat(all_predictions)
```

To really understand how well the model performs, we look at three important things: accuracy, precision, and recall. So let's imagine a case where our model made ***100 predictions*** on whether Tesla’s stock price will go up or not, and here’s how it did:

- **True Positives (TP)**: 40 – These are cases where the model correctly said the price would go up, and it did.  
- **True Negatives (TN)**: 45 – These are when the model predicted no price increase, and it was also right.  
- **False Positives (FP)**: 5 – These are predictions where the model said the price would increase, but it didn’t.  
- **False Negatives (FN)**: 10 – These are when the model missed an actual increase and predicted it wouldn’t go up.



### **1. Accuracy**

**Formula**:  
Accuracy = (True Positives + True Negatives) / Total Predictions  

**Calculation**:  
Accuracy = (40 + 45) / 100 = 0.85 → **85%**

**What It Means**:  
This tells us that the model was right 85 times out of 100, no matter whether the price went up or didn’t. It gives us a general feel about how well it performs overall. But it doesn’t tell us much about what kind of mistakes it makes.



</td>

<td style="width: 48%; text-align: justify; vertical-align: top; padding-left: 20px;">

### **2. Precision**

**Formula**:  
Precision = True Positives / (True Positives + False Positives)

**Calculation**:  
Precision = 40 / (40 + 5) = 0.89 → **89%**

**What It Means**:  
Precision focuses on how reliable the model is when it says the price will go up. So in this case, out of every time the model said “yes, buy it, it will go up,” it was right 89% of the time. That’s really important for us because if we act on its “buy” signal, we don’t want to be misled. A false positive here could mean making a bad move in the market.

This is also why we lean towards high precision—because a wrong “yes” can cost us money. But a missed “yes” (which is a false negative) doesn’t actually make us lose money—it’s just an opportunity we didn’t take.



### **3. Recall**

**Formula**:  
Recall = True Positives / (True Positives + False Negatives)

**Calculation**:  
Recall = 40 / (40 + 10) = 0.80 → **80%**

**What It Means**:  
Recall tells us how many actual price increases the model successfully caught. So, out of 50 real upward moves, it got 40 of them right. That’s 80%. So yeah, it missed 10 chances, but it still managed to catch most of them. If we want to make sure we don’t miss chances to buy, then this is the number we want to be high.

### Comparison between multiple models


```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.linear_model import LogisticRegression
from xgboost import XGBClassifier
from sklearn.svm import SVC
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Initialize classification models
models = {
    'Decision Tree': DecisionTreeClassifier(),
    'Random Forest': RandomForestClassifier(),
    'Extra Trees': ExtraTreesClassifier(),
    'XGBoost': XGBClassifier(),
}

precisions = {}
confusion_matrices = {}
classification_reports = {}

# Train and evaluate each model
for model_name, test_model in models.items():
    # Train the model
    test_model.fit(train[predictors], train["Did_it_increase (Target)"])


    # Make predictions
    test_preds = test_model.predict(test[predictors])

    # Evaluate accuracy
    precision = precision_score(test["Did_it_increase (Target)"], test_preds)


    precisions[model_name] = precision

    # Generate confusion matrix
    cm = confusion_matrix(test["Did_it_increase (Target)"], test_preds)
    confusion_matrices[model_name] = cm

    # Generate classification report
    report = classification_report(test["Did_it_increase (Target)"], test_preds)
    classification_reports[model_name] = report

# Plot all accuracies
# Sort accuracies from highest to lowest
sorted_precisions = {k: v for k, v in sorted(precisions.items(), key=lambda item: item[1], reverse=True)}

# Plot all accuracies
plt.figure(figsize=(10, 7))
bars = plt.bar(sorted_precisions.keys(), sorted_precisions.values(), color='skyblue')
plt.title('Model Precisions')
plt.xlabel('Model')
plt.ylabel('Precision')
plt.ylim(0, 1)

# Annotate each bar with the accuracy value as a percentage
for bar, accuracy in zip(bars, sorted_precisions.values()):
    plt.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.02, f'{accuracy*100:.2f}%',
             ha='center', va='bottom', fontsize=10)

plt.show()

# Plot all confusion matrices
for model_name, cm in confusion_matrices.items():
    plt.figure(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['No', 'Yes'], yticklabels=['No', 'Yes'])
    plt.title(f'{model_name} - Confusion Matrix')
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.show()

# Print all classification reports
for model_name, report in classification_reports.items():
    print(f'{model_name} Classification Report:\n{report}')
```

After breaking down both **precision** and **recall**, it makes sense to also talk about the **F1-score**, which acts as a perfect middle ground. It combines precision and recall into one single number so we can judge performance more completely, especially when the data is not totally balanced. The formula is:

**F1-score = 2 × (Precision × Recall) / (Precision + Recall)**

So in our earlier example, with a precision of 89% and recall of 80%, our F1-score would be about 84%. This shows the model is doing quite well at giving reliable yes-signals while still catching most of the real yeses. That mix of reliability and completeness is really valuable, especially when we don’t want to chase every signal but want the ones we follow to be worth it.

```python
pred_end_date = datetime.datetime.today() + datetime.timedelta(days=90) #predicting 90 days ahead


def monte_carlo(pred_end_date, df=df, iterations=5000, plot=True):
    # Ensure pred_end_date is timezone-aware to match df.index
    if df.index.tz is not None:  # Check if df index is timezone-aware
        pred_end_date = pd.Timestamp(pred_end_date).tz_localize(df.index.tz)
        #otherwise you get a TypeError: Start and end cannot both be tz-aware with different timezones
    """
    Performs a Monte Carlo simulation to predict future stock prices.
    
    Args:
    pred_end_date (datetime): End date for prediction.
    df (pd.DataFrame): Historical stock data.
    iterations (int): Number of simulation iterations.
    plot (bool): Whether to plot the simulation results.
    
    Returns:
    tuple: DataFrame of simulated price paths and final values.
    """

    pred_end_date = pred_end_date
    forecast_dates = [d if d.isoweekday() in range(1, 6) else np.nan for d in pd.date_range(df.index.max(), pred_end_date)]
    intervals = len(forecast_dates)
    iterations = iterations
    #Preparing log returns from data
    log_returns = np.log(1 + df['Close'].pct_change())

    #Setting up drift and random component in relation to asset data
    u = log_returns.mean()
    var = log_returns.var()
    drift = u - (0.5 * var)
    stdev = log_returns.std()
    daily_returns = np.exp(drift + stdev * norm.ppf(np.random.rand(intervals, iterations)))

    #Takes last data point as startpoint point for simulation

    global price_list
    price_list = np.zeros_like(daily_returns)
    price_list[0] = df['Close'].values[-1]
    #Applies Monte Carlo simulation in asset
    for t in range(1, intervals):
        price_list[t] = price_list[t - 1] * daily_returns[t]

    forecast_df = pd.DataFrame(price_list)

    if plot:
        forecast_df.plot(figsize=(10,6), legend=False, title = "{} Simulated Future Paths".format(iterations))

    end_values_df = forecast_df.tail(1)

    return forecast_df, end_values_df

forecast_df, end_values_df = monte_carlo(pred_end_date)
```

```python
def plot_norm_hist(s, vline = True, title= True):
    """
    Plots a histogram of final price distribution with a normal curve overlay.

    Args:
        s (pd.Series): Final simulated prices.
        vline (bool): Whether to draw vertical lines at ±0.67σ.
        title (bool): Whether to display the title with mean and standard deviation.

    Returns:
        None
    """
      
    mu, sigma = np.mean(s), np.std(s) # mean and standard deviation

    count, bins, ignored = plt.hist(s, 30, density=True)
    plt.plot(bins, 1/(sigma * np.sqrt(2 * np.pi)) *
               np.exp( - (bins - mu)**2 / (2 * sigma**2) ),
         linewidth=2, color='r')

    if vline:
        lline = -.67*sigma + mu
        uline = .67*sigma + mu
        plt.axvline(lline, color='g')
        plt.axvline(uline, color='g')

    if title:
        plt.title("Final price distribution\n Mean: ${:.02f} and Standard Deviation: {:.02f}".format(mu, sigma))

    return plt.show()

s = end_values_df.iloc[-1]
plot_norm_hist(s, vline=True, title=True)
```

```python
def predict_trade_tomorrow(test, predictors, model):
    """
    Predicts whether it is a good time to trade today based on the model's probability prediction.

    Args:
        test (pd.DataFrame): The input data containing the relevant features for prediction.
        predictors (list): The list of column names in `test` that will be used as features for the model.
        model (sklearn-like model): A trained model that has a `.predict_proba()` method to return probabilities.

    Returns:
        None: Prints a message indicating whether it is advisable to trade today.
            - "I think you should trade today" if the model's probability of trading is ≥ 0.6.
            - "It is better to not trade today" otherwise.
    """
    if model.predict_proba(test[predictors][-1:])[:,1] >=.6:
        return print("I think you should trade today")
    else:
        return print("It is better to not trade today")
```

```python
def predict_investment(test, end_values_df):
    """
    Predicts whether it is a good time to invest based on a comparison of the mean end values and the current closing price.
    
    Args:
        test (pd.DataFrame): The input data containing the 'Close' column with current closing prices.
        end_values_df (pd.DataFrame): A DataFrame containing historical or calculated end values. The mean of the last row is used for comparison.
    
    Returns:
        None: Prints a message indicating whether it is advisable to invest.
            - "I think you should invest" if the mean of the last row in `end_values_df` is greater than or equal to the last closing price.
            - "It is better to not invest" otherwise.
    """

    if np.mean(end_values_df.iloc[-1]) >= test['Close'].values[-1]:
        return print("I think you should invest")
    else:
        return print("It is better to not invest")
```
